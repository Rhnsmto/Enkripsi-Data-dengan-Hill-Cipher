```python
import numpy as np

# Konversi huruf ke angka: A=0, B=1, ..., Z=25
def text_to_numbers(text):
    return [ord(char.upper()) - ord('A') for char in text]

# Konversi angka ke huruf: 0=A, 1=B, ..., 25=Z
def numbers_to_text(numbers):
    return ''.join([chr(num % 26 + ord('A')) for num in numbers])

# Fungsi enkripsi Hill Cipher
def hill_cipher_encrypt(plaintext, key_matrix):
    # Pastikan panjang plainteks genap (untuk matriks 2x2)
    if len(plaintext) % 2 != 0:
        plaintext += 'X'  # tambahkan padding jika perlu

    # Konversi ke angka
    nums = text_to_numbers(plaintext)

    # Bagi ke blok 2 angka
    blocks = [nums[i:i+2] for i in range(0, len(nums), 2)]

    ciphertext_nums = []

    for block in blocks:
        # Kalikan matriks kunci dengan blok, lalu mod 26
        vector = np.array(block).reshape(2, 1)
        result = np.dot(key_matrix, vector) % 26
        ciphertext_nums.extend(result.flatten())

    # Konversi hasil ke huruf
    ciphertext = numbers_to_text(ciphertext_nums)
    return ciphertext

# ==== Contoh Penggunaan ====

# Matriks kunci (harus invertible di mod 26)
key_matrix = np.array([[3, 3],
                       [2, 5]])

plaintext = "RAIHAN"
ciphertext = hill_cipher_encrypt(plaintext, key_matrix)

print("Plaintext:", plaintext)
print("Ciphertext:", ciphertext)
```
